 2. 'DistanceFromHome' and most job-related features: There is not always a strong correlation, which may be counter to expectations that living farther would influence work outcomes.

```python
import matplotlib.pyplot as plt

# Filter only employees with Attrition == 'Yes'
df_yes = df_merged[df_merged['Attrition'] == 'Yes']

# Calculate total number of 'Attrition = Yes' employees
total_attrition_yes = len(df_yes)

# Calculate counts per Education among those who left
attrition_yes_counts = (
    df_yes
    .groupby('Education')
    .size()
    .reindex(sorted(df_merged['Education'].unique()), fill_value=0)
)

# Percentage for each education level out of all Attrition = Yes population
attrition_yes_percent = 100 * attrition_yes_counts / total_attrition_yes

# Plot
plt.figure(figsize=(8,5), dpi=100)
bars = plt.bar(
    attrition_yes_percent.index.astype(str), 
    attrition_yes_percent.values,
    color='#feb24c',
    edgecolor='black',
    alpha=0.85
)

plt.title("Percent of worker who left by Education Level\n(Relative to All Left Employees)", fontsize=15, weight='bold')
plt.xlabel('Education Level', fontsize=12)
plt.ylabel('percent of workers who left from all workers', fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

# Annotate each bar with its percentage
for bar in bars:
    height = bar.get_height()
    plt.annotate(f"{height:.2f}%", 
                 (bar.get_x() + bar.get_width() / 2, height), 
                 ha='center', va='bottom', 
                 fontsize=10, fontweight='bold', color='#222222'
                )

plt.tight_layout()
plt.show()
```

* The bar chart shows that employees with an Education level of 3 make up the largest group among both those who have left the company (Attrition = Yes) and those who have stayed (Attrition = No). 
* However, the distribution of attrition is relatively similar across all education levels, indicating that education level alone does not have a strong or obvious correlation with employee attrition in this dataset.


```python
# Create a pivot table with employee counts by Attrition and JobLevel
pivot_table = df_merged.pivot_table(
    values="EmployeeNumber",
    index="Attrition",
    columns="JobLevel",
    aggfunc='count',
    fill_value=0
)

# Transpose to have JobLevel as rows for percentage calculations
pivot_table_T = pivot_table.T

# Calculate percentage by JobLevel (row-wise total per Job Level = 100%)
pivot_table_perc = pivot_table_T.div(pivot_table_T.sum(axis=1), axis=0) * 100

import matplotlib.pyplot as plt

# Define bar colors: first two bars (Job Level 1, both attrition values) colorful, others grey faded
# Change the green bar (#66c2a5) to blue (#1f77b4)
job_levels = list(pivot_table_perc.index)
attrition_categories = list(pivot_table_perc.columns)
num_attrition = len(attrition_categories)
bar_colors = []

colorful_bar_colors = ['#1f77b4', '#fc8d62']  # Blue & orange
grey_bar = '#dddddd'

for i, jl in enumerate(job_levels):
    group_colors = []
    if i == 0:  # JobLevel 1
        group_colors = colorful_bar_colors[:num_attrition]
    else:
        group_colors = [grey_bar] * num_attrition
    bar_colors.append(group_colors)

# bar_colors needs to be transposed for DataFrame color param (list of color for each bar)
# In grouped barplot, color expects a list with length = number of columns
# We'll have to flatten the color list by column-wise ordering
flattened_colors = []
for col_idx in range(num_attrition):
    for row_idx in range(len(job_levels)):
        flattened_colors.append(bar_colors[row_idx][col_idx])

# Set figure size and style
fig, ax = plt.subplots(figsize=(10, 6), dpi=100)

width = 0.8
bar_width = width / num_attrition
indices = np.arange(len(job_levels))

bars = []
for i, attr in enumerate(attrition_categories):
    # Bar color for this attr and each joblevel
    this_colors = [bar_colors[j][i] for j in range(len(job_levels))]
    # Heights in correct order
    this_heights = pivot_table_perc[attr]
    # bar positions
    bar_pos = indices + i * bar_width - width/2 + bar_width/2
    bars.append(ax.bar(
        bar_pos,
        this_heights,
        width=bar_width,
        color=this_colors,
        edgecolor='black',
        alpha=0.85,
        label=str(attr)
    ))

# Professional labels and title
ax.set_title("Attrition Percentage by Job Level", fontsize=16, weight='bold')
ax.set_xlabel("Job Level", fontsize=13)
ax.set_ylabel("Percentage of Employees (%)", fontsize=13)
ax.set_xticks(indices)
ax.set_xticklabels(job_levels, fontsize=11)
ax.set_yticks(np.arange(0, 101, 10))
ax.tick_params(axis='y', labelsize=11)
ax.set_ylim(0, 100)

# Legend formatting
ax.legend(title='Attrition', bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=11, title_fontsize=12, frameon=True)

plt.tight_layout()

# Annotate value labels on each bar (percentage)
for bar_group in bars:
    for bar in bar_group:
        height = bar.get_height()
        if height > 0:
            ax.annotate(f'{height:.1f}%',
                        (bar.get_x() + bar.get_width() / 2, height),
                        ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.show()
```

* The above graph reveals that employee attrition rates are highest among those with relatively few years at the company, for both males and females. The number of people leaving drops considerably as tenure increases, especially after the first couple of years. This suggests that relatively new employees are at greater risk of leaving, and retention strategies may be especially important for early-tenure staff, regardless of gender.

```python
# Visualize pivot table (Attrition x JobRole) as bar chart showing percentage and fix legend/color mapping
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch

pivot_jobrole = df_merged.pivot_table(
    values="EmployeeNumber",
    index="Attrition",
    columns="JobRole",
    aggfunc='count',
    fill_value=0
)

# Convert counts to percentages per JobRole
jobrole_totals = pivot_jobrole.sum(axis=0)
pivot_jobrole_percent = pivot_jobrole.divide(jobrole_totals, axis=1) * 100

jobroles = list(pivot_jobrole_percent.columns)
attrition_levels = list(pivot_jobrole_percent.index)
n_attr = len(attrition_levels)
n_jobs = len(jobroles)
bar_width = 0.8 / n_attr  # So the total group width is ~0.8
indices = np.arange(n_jobs)

# Find sales rep index
try:
    sales_rep_idx = jobroles.index('Sales Representative')
except ValueError:
    sales_rep_idx = None

# Define color schemes
faded_grey = ['#cccccc', '#b3b3b3']
colorful_colors = ['#8da0cb', '#fc8d62']  # For 2 attrition statuses

fig, ax = plt.subplots(figsize=(14, 7), dpi=100)
bars = []

# We'll track the legend patches/labels
legend_patches = []

for a, attr in enumerate(attrition_levels):
    heights = [pivot_jobrole_percent.iloc[a, j] for j in range(n_jobs)]
    # Create color list for this attrition category
    colors_this = []
    for j in range(n_jobs):
        if sales_rep_idx is not None and j == sales_rep_idx:
            colors_this.append(colorful_colors[a])
        else:
            # fallback for arbitrary number of attrition categories
            if a < len(faded_grey):
                colors_this.append(faded_grey[a])
            else:
                # Generate/assign a grey for further bars
                greyval = 204 - 25*a
                if greyval < 0: greyval = 160
                colors_this.append(f'#{greyval:02x}{greyval:02x}{greyval:02x}')
    bar_pos = indices + a * bar_width - bar_width*(n_attr-1)/2
    bar_container = ax.bar(
        bar_pos,
        heights,
        width=bar_width,
        color=colors_this,
        edgecolor='black',
        alpha=0.88,
        label=None  # We'll handle legend manually
    )
    bars.append(bar_container)
    # Add only one patch for faded and one for colorful per attrition category
    if a < len(faded_grey):
        # For generic (non-sales rep) bars
        legend_patches.append(Patch(facecolor=faded_grey[a], edgecolor='black', label=f"{attr} (Other Job Roles)"))
    if sales_rep_idx is not None:
        legend_patches.append(Patch(facecolor=colorful_colors[a], edgecolor='black', label=f"{attr} (Sales Representative)"))
    # Add percentage label above colorful bar(s)
    if sales_rep_idx is not None:
        bar = bar_container[sales_rep_idx]
        height = bar.get_height()
        if height > 0:
            ax.annotate(f'{height:.1f}%',
                        (bar.get_x() + bar.get_width() / 2, height),
                        ha='center', va='bottom', fontsize=13, fontweight='bold', color='black')

ax.set_title("Attrition Percentage by Job Role", fontsize=16, weight='bold')
ax.set_xlabel("Job Role", fontsize=13)
ax.set_ylabel("Percentage of Employees (%)", fontsize=13)
ax.set_xticks(indices)
ax.set_xticklabels(jobroles, rotation=25, ha='right', fontsize=11)
ax.set_yticks(np.arange(0, 101, 10))
ax.tick_params(axis='y', labelsize=11)
ax.set_ylim(0, 100)
ax.legend(handles=legend_patches, title='Attrition x JobRole', bbox_to_anchor=(1.02, 1),
          loc='upper left', fontsize=11, title_fontsize=12, frameon=True)
plt.tight_layout()
plt.show()
```

* Employees in the Sales department had the highest attrition rate compared to other departments, indicating a greater tendency for workers to leave this department.




 ## Significant differences in the following parameters

* Age: Employees who left are mainly in the second half of their 30s (0.75 percentile left VS stayed: 39 VS 43)
* job level: 60% of the employees who lef were in entry-level positions
* Department: Sales department leads in num of employees who left
* Job Role: Mainly employees left positions of: Laboratory Technician, Sales Excutive and Research Scientist 
* Monthly Rate: A small difference between the 2 groups (mean_left= 14,265$ VS mean_stayed= 14,559$)




**Summary:**
Analysis of the attrition data reveals that the highest rates of employee turnover occur within the Sales department. Employees who left tended to be in their late 30s, and a significant proportion held entry-level positions. Specific job roles with the greatest attrition include Laboratory Technician, Sales Executive, and Research Scientist. Differences in monthly rate between employees who left and those who stayed are minor.

**Recommendations:**
1. **Targeted Retention Strategies:** Focus on supporting employees in entry-level roles, especially in the Sales department and the high-attrition job roles. Implement mentorship programs or provide clear career advancement paths to boost retention.
2. **Review Compensation & Benefits:** Although differences in pay were small, periodic review of compensation, especially for sales staff, may help address attrition.
3. **Work Environment Enhancements:** Consider evaluating the workplace culture and workload in sales and technical roles to identify and address potential sources of dissatisfaction.
4. **Exit Interview Analysis:** Systematically analyze exit interview data for these key groups to identify specific drivers of turnover and inform targeted interventions.
